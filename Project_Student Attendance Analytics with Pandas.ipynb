## Student Attendance Analytics with Pandas

```python
import pandas as pd
import numpy as np
```

#### Section 0: Data Loading

```python
df = pd.read_excel("aids_attendance.xlsx", sheet_name="Student Summary")
df
```

#### Section 1: Data Inspection

```python
print("First 10 rows:")
display(df.head(10))
print("Shape:", df.shape)
print("Data types:")
print(df.dtypes)
```

```python
print("Null values by column:")
print(df.isna().sum())
print("Literal 'Never' values in Last Present Date:", df["Last Present Date"].eq("Never").sum())
```

```python
df = df.rename(columns={
    "Roll No": "roll_no",
    "Student Name": "student_name",
    "Subject": "subject",
    "Total Classes": "total_classes",
    "Present": "present",
    "Absent": "absent",
    "Excused": "excused",
    "Attendance %": "attendance_pct",
    "Status": "status",
    "Classes Needed (75%)": "classes_needed_75_pct",
    "Last Present Date": "last_present_date",
    "Consecutive Absences": "consecutive_absences",
})
print(df.columns.tolist())
```

```python
df["attendance_pct"] = df["attendance_pct"].str.rstrip("%").astype(float)
df["last_present_date"] = pd.to_datetime(
    df["last_present_date"].replace("Never", pd.NA), errors="coerce"
)
print(df[["attendance_pct", "last_present_date"]].dtypes)
display(df[["attendance_pct", "last_present_date"]].head())
```

#### Section 2: Filtering & Selection

```python
critical_students = df.loc[df["status"].eq("Critical")].sort_values("attendance_pct")
display(critical_students)
```

```python
fully_absent = df.loc[df["present"].eq(0), ["roll_no", "student_name", "consecutive_absences"]]
display(fully_absent)
```

```python
safe_students = df.loc[df["status"].eq("Safe")]
print("Safe students:")
display(safe_students)
print("First 15 rows, columns 2-6:")
display(df.iloc[:15, 1:6])
```

```python
high_absence_students = df.loc[
    df["consecutive_absences"].ge(5),
    ["student_name", "status", "last_present_date"],
]
high_absence_students
```

### Section 3: Grouping & Aggregation 

```python
status_summary = df.groupby("status").agg(
    student_count=("roll_no", "size"),
    average_attendance_pct=("attendance_pct", "mean"),
    max_consecutive_absences=("consecutive_absences", "max"),
)
status_summary
```

```python
conditions = [
    df["attendance_pct"].ge(90),
    df["attendance_pct"].ge(75),
    df["attendance_pct"].ge(60),
]
labels = ["Excellent", "Good", "Average"]
df["attendance_category"] = np.select(conditions, labels, default="Poor")
category_counts = df.groupby("attendance_category").size().sort_index()
category_counts
```

```python
overall_totals = df.agg(
    total_classes_conducted=("total_classes", "sum"),
    total_present=("present", "sum"),
    total_absent=("absent", "sum"),
)
overall_totals
```

```python
status_attendance_summary = df.groupby("status").agg(
    present_min=("present", "min"),
    present_max=("present", "max"),
    present_mean=("present", "mean"),
    absent_sum=("absent", "sum"),
)
status_attendance_summary
```

### Section 4: Feature Engineering & Transformation

```python
df["risk_score"] = (df["absent"] * 0.4) + (df["consecutive_absences"] * 0.6)
top_10_risk = df.nlargest(10, "risk_score")
top_10_risk[["roll_no", "student_name", "absent", "consecutive_absences", "risk_score"]]
```

```python
df["classes_needed_numeric"] = (
    df["classes_needed_75_pct"].str.extract(r"(\d+)")[0].fillna("0").astype(int)
)
df[["classes_needed_75_pct", "classes_needed_numeric"]].head(10)
```

```python
df["attendance_flag"] = df["attendance_pct"].apply(
    lambda attendance: "Defaulter" if attendance < 75 else "Safe"
)
df[["student_name", "attendance_pct", "attendance_flag"]].head(10)
```

```python
df["attendance_range"] = pd.cut(
    df["attendance_pct"],
    bins=[-np.inf, 50, 75, 90, np.inf],
    labels=["Critical", "Warning", "Good", "Excellent"],
    right=True,
)
df[["student_name", "attendance_pct", "attendance_range"]].head(10)
```

### Section 5: Advanced Analysis 

```python
status_category_pivot = pd.pivot_table(
    df,
    index="status",
    columns="attendance_category",
    values="roll_no",
    aggfunc="count",
    fill_value=0,
)
status_category_pivot
```

```python
df["base_name"] = df["student_name"].str.extract(r"^\s*([A-Za-z]+)")[0].str.upper()
repeated_base_names = df.loc[df["base_name"].duplicated(keep=False)].sort_values("base_name")
repeated_base_names[["roll_no", "student_name", "base_name"]]
```

```python
Student_Details = df[["roll_no"]].drop_duplicates().copy()
Student_Details["branch"] = "AI&DS"
Student_Details["semester"] = "Not specified"
merged_students = df.merge(Student_Details, on="roll_no", how="inner")
merged_students
```

```python
critical_students.to_csv("critical_students.csv", index=False)
status_summary = df.groupby("status").agg(
    student_count=("roll_no", "size"),
    average_attendance_pct=("attendance_pct", "mean"),
    max_consecutive_absences=("consecutive_absences", "max"),
).reset_index()
status_summary.to_csv("attendance_summary_by_status.csv", index=False)
top_10_risk.to_csv("top_10_students_by_risk_score.csv", index=False)
print("Exported critical_students.csv, attendance_summary_by_status.csv, and top_10_students_by_risk_score.csv")
```
